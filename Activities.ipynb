# Hands-On Activity 1
## Environment Setup Laboratory

**Notice:** This is the environment every future week's activities will build on top of — do not skip steps. If you were able to follow along last laboratory session, you can proceed with step 5

---

### Part 1 — Install Core Tools

1. Install **Node.js LTS** from https://nodejs.org (choose the "LTS" button, not "Current").
2. Install **Visual Studio Code** from https://code.visualstudio.com.
3. Install **Git** from https://git-scm.com.
4. Create a free account on https://github.com if you don't already have one.
5. Verify everything installed correctly:
   ```bash
   node -v
   npm -v
   git --version
   ```
### Part 2 — VS Code Extensions

Install these extensions inside VS Code (Extensions panel, left sidebar):
- **ESLint**
- **Prettier – Code formatter**
- **GitLens**

### Part 3 — Scaffold the Project

1. Open a terminal in the folder where you keep your school projects.
2. Run:
   ```bash
   npx create-next-app@latest studyboard
   ```
   When prompted, choose:
   - TypeScript → **Yes**
   - ESLint → **Yes**
   - Tailwind CSS → **Yes**
   - `src/` directory → **No**
   - App Router → **Yes**
   - Import alias → accept default (`@/*`)
3. Move into the project and start the dev server:
   ```bash
   cd studyboard
   npm run dev
   ```
4. Open http://localhost:3000 in your browser. Take a screenshot of the default Next.js welcome page.

### Part 4 — Make Your First Edit

1. Open `studyboard` in VS Code (`code .` from the terminal, or File → Open Folder).
2. Open `app/page.tsx` and replace the main heading text with:
   ```
   Welcome to StudyBoard
   ```
3. Save the file and confirm the browser updates automatically (this is called **hot reload**).
4. Take a screenshot of your updated homepage.

### Part 5 — Git & GitHub

1. Initialize Git inside the `studyboard` folder:
   ```bash
   git init
   ```
2. Confirm a `.gitignore` file exists and that it lists `node_modules` and `.env*` — `create-next-app` generates this automatically. Open it and check.
3. Stage and commit your work:
   ```bash
   git add .
   git commit -m "Initial commit: StudyBoard project scaffolded"
   ```
4. On GitHub, create a **new empty repository** named `studyboard` (do **not** initialize it with a README).
5. Connect and push:
   ```bash
   git remote add origin https://github.com/<your-username>/studyboard.git
   git branch -M main
   git push -u origin main
   ```
6. Refresh your GitHub repository page — confirm your files are there.
